```python
import matplotlib.pyplot as plt
from matplotlib.patches import Rectangle, FancyArrowPatch

def visualize_real_model():
    fig, ax = plt.subplots(figsize=(10, 3))

    blocks = [
        "AtomEdgeInteraction",
        "Atom Preprocess x8",
        "ChebConv(128->128, K=10)",
        "ChebConv(128->128, K=16)",
        "Postprocess x2",
        "Output Layer"
    ]

    x_start = 0.0
    width = 1.4
    height = 0.6
    spacing = 0.8

    for i, block_name in enumerate(blocks):
        rect = Rectangle((x_start, 0), width, height, fill=True, edgecolor='black')
        ax.add_patch(rect)
        ax.text(x_start + width/2, height/2, block_name, ha='center', va='center')
        if i < len(blocks) - 1:
            arrow = FancyArrowPatch((x_start + width, height/2),
                                    (x_start + width + spacing/2, height/2),
                                    arrowstyle='->')
            ax.add_patch(arrow)
        x_start += width + spacing

    ax.set_xlim(-0.5, x_start)
    ax.set_ylim(-0.5, height + 0.5)
    ax.axis('off')
    plt.savefig("real_model_diagram.png", dpi=150, bbox_inches='tight')
    print("Diagram saved as 'real_model_diagram.png'.")

if __name__ == "__main__":
    visualize_real_model()
```
